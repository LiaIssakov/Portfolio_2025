# Task 3: Chains of matrices

Given a list of matrices $[F_1, F_2, \dots, F_n]$, corresponding to linear transformations $f_1, f_2, \dots f_n$, compute $f_n(f_{n-1}(\dots f_2(f_1(v))))$ for some vector $v$.

> I'm fishing for a basic solution using a for loop! Assuming the matrices are large, the for loop will not slow things down.

Verify it works correctly!

```python
def apply_composition(Fs, v):
  for matrix in Fs:
    v = matrix @ v
  return v
```

```python
assert np.allclose(apply_composition([R, S], [2,0]), [0,1]) # example from above
```

# Task 4:

Implement your own matrix-matrix multiplication

> You can assume square matrices at first, it's slightly easier.

```python
# m is the number of elements in our matrix (n^2) -- input size
# O(n^3) but what about m? O(n^3) = O(sqrt(m)^3) = )(m^(3/2)) = O(m^1.5)
def mul(A, B):
    p, q = A.shape
    q2, r = B.shape
    assert q == q2

    R = np.zeros((p, r)) # let's just fill it up

    # finish me using for loops (no np.dot, @ etc)
    for i in range(p):
      for j in range(r):
        for k in range(q):
          R[i, j] += A[i, k] * B[k, j]

    return R
```

```python
A = np.random.rand(3, 5)
B = np.random.rand(5, 8)
assert np.allclose(mul(A, B), A@B)
```

# Optional Task 5: More efficient matrix chain

Suppose you know that someone will use the above function many times using the same list Fs, but **many different vectors**, which we encode as the rows in an array $X$. Can you speed things up compared to running the above function for each input?

> More advanced python thing: could you not take X, but instead allow the used to easily and efficiently apply any single vector they want (without them having to know about matrix multiplication, etc)? (As usual I'm fishing for a particular solution and I'm being vague about what I really want...)

Usage should be something like:
R = prepare(Fs)
res = R(X)

```python
def apply_composition_on_many_vectors(Fs, X):
    transformation_product = Fs[0]
    for matrix in Fs[1:]:
      transformation_product = matrix @ transformation_product
    return transformation_product @ np.nsarray(X)
```

```python
assert np.allclose(apply_composition_on_many_vectors([R, S], [[2,0], [2,0]]), [[0,1], [0,1]])
```
