# Image data

We will be using various standard image datasets used in machine learning.

The data will download automatically and come as numpy arrays

Note that the data is split into
- *train and *test data (for training and test)
- also into x_ and y_ part (data, labels)

> This can be confusing: x means the images and y means the labels! It's a standard convention used in data science. But it's easy to assume these are the x and y coordinates of images -- they are not.

So overall we get two pairs, which we can directly 'unpack' into 4 variables, see below.

> Again: x_train/x_test are images formated as numpy arrays. y_train/y_test are the corresponding (numerical) labels also formated as numpy arrays.

For the mnist dataset each image shows a hand-written (decimal) digit, the corresponding label is the numerical value of the drawn digit -- so just an integer between 0 and 9.

# How can such data be used?


## Unsupervised learning (without using labels):
- clustering: using only the images, try to find a small number of representative images. For mnist it could be just 10 images, each showing a different digit. Or maybe there are significant variations within each class -- so there should be more clusters?

- dimensionality reduction (especially for data visualization): take the images and plot each image as a point in 2D so that points that are close together represent images which are visually similar.

## Supervised learning (using labels):
- classification: you first train a model using x_train (images) and y_train (labels); then you try to predict labels on unseen images. So if I give you a new image showing a hand-written six, you should say '6'.
- generation: you again train on x_train and y_train and then try to generate images based on their labels. So if I say '6' you should give me a new image of a hand-drawn six.
